```python
str1=input("enter first string:")
str2=input("enter second string:")
str3=str1+" "+str2
print("concatenated string is:",str3)
print("repetation of str1:",str1*9)
print("upper case of str1:",str1.upper())
print("lower case of str2:",str2.lower())
print("length of str1:",len(str1))
print("capitalize:",str1.capitalize())
print("center:",str1.center(10))
print("repalce:",str2.replace("a","@"))
print(f"My first name is {str1} and second name is {str2}")
print("Removing unwanted space:",str1.strip(" "))
```
