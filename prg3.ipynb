```python
import matplotlib.pyplot as plt
import numpy as np
#plot1
x=np.array([10,50])
y=np.array([100,500])
plt.title("Title 1")
plt.xlabel("xlabel",color="blue")
plt.ylabel("ylabel",color="blue")
plt.subplot(2,3,1)
plt.scatter(x,y)

#plot 2
ypoints = np.array([4,8,2,7])
xpoints = np.array([7,3,9,6])
plt.subplot(2,3,2)
plt.plot(ypoints,marker='*',ms=10,color='orange')
plt.plot(xpoints,linestyle='dotted',color='blue')

#plot3


ypoints = np.array([4,10,2,9])
plt.subplot(2,3,3)
plt.plot(ypoints,marker='o',ms=10,color='blue',mfc='yellow',mec='black')

#plot4
x=np.array([80,85,90,95,100,105,110,115,120,125])
y=np.array([240,250,260,270,280,290,300,310,320,330])
plt.subplot(2,3,4)
plt.title("Title 2")

plt.plot(x,y)
plt.grid()
#plot5

x = np.array([5,7,8,7,2,17,7])
y = np.array([99,86,87,88,111,89,100])
colors=np.array(['red','green','blue','magenta','yellow','pink','black'])
sizes=np.array([200,40,50,20,30,50,500])
plt.subplot(2,3,5)
plt.scatter(x,y,c=colors,s=sizes,alpha=0.5)

#plot6                                                                                                                                                                           

x = np.array(["A", "B", "C", "D"])
y = np.array([3, 8, 1, 10])
plt.subplot(2,3,6)
plt.bar(x,y,color='red',width=0.1)


plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np



y = np.array([35, 20, 20, 25])
mylabels = ["Apples", "Bananas", "Cherries", "Dates"]
myexplode = [0.1, 0, 0., 0]
mycolors=["red","green","blue","yellow"]

plt.pie(y,labels=mylabels,startangle=90,explode=myexplode,shadow=True,colors=mycolors)
plt.legend()


```
